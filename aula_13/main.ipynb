### Listando e buscando musicas
- Exibe os objetos existentes
- Buscar procura uma parte do nome ou do cantor
- Filtrar seleciona músicas com determinado gênero ou ano

```python
#c2.listar_musicas()

# filtros por nome
for musica in c2.filtrar_por_nome('azul'):
    print(musica)
# filtro por cantor
for musica in c2.filtrar_por_cantor('bruno'):
    print(musica)
# filtro por genero
for musica in c2.filtrar_por_genero('trap'):
    print(musica)
# filtro por duração
for musica in c2.filtrar_por_duracao(2,4):
    print(f'música: {musica}, duração: {musica.duracao}')
# filtrar por ano
for musica in c2.filtrar_por_ano(1999):
    print(f'musica: {musica}, ano: {musica.ano}')
```

# Removendo a Música do catálogo
