### Métodos
- Funções definidas dentro de uma classe, que representam operações ou comportamentos associados aos seus objetos

Música:

Exibe suas informações:
- Mostra o título, o artista, a duração e avaliação

Registrar uma reprodução:
- Aumentar a quantidade de reproduções



```python
class Musica:
    def __init__(self, titulo):
        self.titulo = titulo
        self.reproducoes = 0

    def reproduzir(self):
        self.reproducoes +=1
```

### Prática:

| Atributos | Tipo |
|--- | --- |
| nome    | str |
| cantor  | str |
| genero  | str |
| duracao | int |
| ano     | int |


```python
class Musica:
    def __init__ (self,nome,cantor,genero,duracao,ano):
        self.nome= nome
        self.cantor= cantor
        self.genero= genero
        self.duracao= duracao
        self.ano= ano

        # atributos de controle:

        self.reproducoes = 0
        self.tocando =False #booleana


    # Padronize os nomes das funções
    # Evite colocar o nome da função com mesmo nome do atributo (na hora de instanciar, chama a função e atributo)

    def mostrar_nome(self):
        print(f'nome da música: {self.nome}')
    def mostrar_cantor(self):
        print(f'cantor: {self.cantor}')
    def mostrar_genero(self):
        print(f'genero: {self.genero}')
    def mostrar_duracao(self):
        print(f'duracao: {self.duracao}')
    def mostrar_ano(self):
        print(f'ano: {self.ano}')


    #metodo de exibir todas as informações bases da música
    def mostrar_faixa(self):
        self.mostrar_nome()
        self.mostrar_cantor()
        self.mostrar_genero()
        self.mostrar_duracao()
        self.mostrar_ano()

    #==========================================================

    def reproduzir(self):
        if self.tocando == False:
            self.tocando = True
            self.reproducoes+=1
            print(f'reproduzindo: {self.nome}')
        else:
            print(f'essa música já está sendo executada')

    def pausar(self):
        if self.tocando == True:
            self.tocando = False
            print('Música pausada')
        else:
            print('A música não está sendo reproduzida')
    
```

```python
m1 = Musica('Caneta Azul', 'Manoel Gomes', 'Brega', 73, 2019)
m1.mostrar_nome()
m1.mostrar_cantor()
m1.mostrar_genero()
m1.mostrar_duracao()
m1.mostrar_ano()

print("\ninfomrações detalhadas: \n")
m1.mostrar_faixa()
```

```python
m1.reproduzir()
print(m1.reproducoes)

m1.reproduzir()

m1.pausar()

m1.reproduzir()
print(m1.reproducoes)
```
