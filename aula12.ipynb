### Encapsulamento
- garantir que a regra seja respeitada msm após a criação do projeto
- @property - modo de leitura


```python
class Musica:
    def __init__(self, nome, cantor, genero, duracao, ano):

        # validação
        if type(duracao) is not int or duracao <=0:
            raise ValueError('O atributo duração deve ser inteiro e positivo')
        if type(ano) is not int or ano <=0:
            raise ValueError('O ano da música deve ser inteiro e positivo')
        if not isinstance(nome,str) or not nome.strip():
            raise ValueError('O nome da música deve ser um texto não vazio')
        if not isinstance(cantor,str) or not cantor.strip():
            raise ValueError('O nome do cantor deve ser un texto não vazio')
        if not isinstance(genero,str) or not genero.strip():
            raise ValueError('O gênero deve ser um texto ñ vazio')
        

        
        self._nome = nome.strip()
        self._cantor = cantor.strip()
        self._genero = genero.strip()
        self._duracao = duracao
        self._ano = ano

        # Atributos de controle
        self.reproducoes = 0
        self.tocando = False
        self.favorito = False
        #++++++++++++++++++++++++++++++++#
        # Propriedades internas (@property)
        @property
        def nome(self):
            return self._nome
        #++++++++++++++++++++++++++++++++#
        @property
        def cantor(self):
            return self._cantor
        #++++++++++++++++++++++++++++++++#
        @property
        def genero(self):
            return self._genero
        #++++++++++++++++++++++++++++++++#
        @property
        def duracao(self):
            return self._duracao
        #++++++++++++++++++++++++++++++++#
        @property
        def ano(self):
            return self._ano
        #++++++++++++++++++++++++++++++++#

    # Métodos para exibir informações
    def mostrar_nome(self):
        print(f"Nome da música: {self._nome}.")

    def mostrar_cantor(self):
        print(f"Cantor(a): {self._cantor}.")

    def mostrar_genero(self):
        print(f"Gênero: {self._genero}.")

    def mostrar_duracao(self):
        print(f"Duração: {self._duracao} segundos.")

    def mostrar_ano(self):
        print(f"Ano: {self._ano}.")

    # Método geral, que reutiliza outros métodos
    def mostrar_detalhe(self):
        self.mostrar_nome()
        self.mostrar_cantor()
        self.mostrar_genero()
        self.mostrar_duracao()
        self.mostrar_ano()
        self.favoritar()
        print(f'Favorito: {'sim' if self._favorito else 'não'}')


    # Métodos que modificam o estado do objeto
    def reproduzir(self):
        if self.tocando == False:
            self.tocando = True
            self.reproducoes += 1
            print(f"Reproduzindo: {self._nome}")
        else:
            print("Essa música já está sendo tocada!")

    def pausar(self):
        if self.tocando == True:
            self.tocando = False
            print("Música pausada!")
        else:
            print("A música não está sendo reproduzida!")

    def favoritar(self):
        if self.favorito == False:
            self.favorito = True
        else:
            print('Já favoritada')

    def desfavoritar(self):
        if self.favorito == True:
            self.favorito = False
        else:
            print('Não favoritada')

    def alterar_nome(self,novo_nome):
        self._nome = novo_nome
        print(f'nome atualizado para {self._nome}')

    def alterar_genero(self,novo_genero):
        self._genero = novo_genero
        print(f'gênero atualizado para {self._genero}')

    def alterar_duracao(self,nova_duracao):
        self._duracao = nova_duracao
        print(f'duracao atualizada para {self._duracao}')
    def alterar_ano(self,novo_ano):
        self._ano = novo_ano
        print(f'atualização do ano da música para {self._ano}')
```
